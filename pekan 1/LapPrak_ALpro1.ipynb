## 2.1 Dasar Teori
Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub membahas dasar penggunaan bahasa pemrograman Python, Visual Studio Code sebagai code editor, serta GitHub sebagai media penyimpanan dan pengelolaan kode program.

  1. Python<br>
 phyton adalah bahasa pemrograman tingkat tinggi (high-level programming language) yang bersifat interpreted, artinya kode dieksekusi baris per baris oleh interpreter tanpa perlu proses kompilasi terpisah seperti pada bahasa C atau Java.
  2. Visual Studio Code
 VS Code adalah source-code editor ringan namun kaya fitur yang dikembangkan oleh Microsoft, digunakan untuk menulis, mengedit, men-debug, dan menjalankan kode dalam berbagai bahasa pemrograman.
  3. Git<br>
 Git adalah sistem kontrol versi terdistribusi (distributed version control system) yang mencatat perubahan pada berkas kode dari waktu ke waktu, memungkinkan pengembang melacak riwayat perubahan dan berkolaborasi.
  4. GitHub<br>
 platform hosting berbasis cloud untuk repository Git, mendukung kolaborasi dan penyimpanan kode secara online, dengan autentikasi melalui HTTPS (token) atau SSH.
  5. Repository<br>
 adalah wadah penyimpanan proyek; commit adalah titik penyimpanan perubahan; branch adalah cabang pengembangan terpisah dari kode utama.
  6. Commit dan Push<br>
 commit adalah proses menyimpan "snapshot" perubahan kode ke dalam riwayat repository lokal. Setiap commit disertai commit message yang menjelaskan perubahan apa yang dilakukan, sehingga riwayat perkembangan proyek dapat dilacak dengan jelas.sedangkan push adalah proses mengirim commit yang tersimpan di repository lokal ke repository remote (misalnya di GitHub), sehingga perubahan tersebut dapat diakses dan dilihat oleh orang lain atau disinkronkan antar perangkat.

### 2.2.1 Program "Claudya Annisa_2611533024"

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini" + "dibuat oleh a\n"
+ "claudya annisa dan 2611533024.")
```

**Penjelasan :**
1.	Bagian pertama, bagian ini adalah teks yang diapit tiga tanda kutip. Karena tidak disimpan ke variabel atau dipanggil, Python menganggapnya sebagai komentar dan tidak ikut dieksekusi/dicetak. Fungsinya hanya sebagai catatan/dokumentasi program (judul tugas, keterangan pertemuan, dll).
2.	Bagian kedua, bagian ini berisi deretan print sederhana. Pada baris pertama, print() mencetak teks Hello World. Pada baris kedua, kode print mencetak baris kosong. Pada baris ketiga, print() mencetak teks Program ini menghasilkan. Pada baris keempat, print() mencetak teks Empat baris output. Pada baris terakhir, berisi karakter escape \\, yang di dalam string Python setiap \\ mewakili satu karakter \ sehingga hasil cetaknya adalah: / \\ // \\\\ /// \\\\\\
3.	Bagian ketiga, bagian ini berisi print() dengan penggabungan string (concatenation). Tanda + menggabungkan tiga potongan string jadi satu kalimat utuh. \n di tengah string berarti “pindah ke baris baru” saat dicetak. Kode ditulis menjadi dua baris fisik hanya untuk kerapian, Python tetap membacanya sebagai satu baris print() karena tanda kurungnya belum ditutup di baris pertama.
