## 2.1 Dasar Teori
Dalam pemrograman Python, operator merupakan salah satu bagian penting yang digunakan untuk melakukan berbagai operasi terhadap data dan variabel. Operator membantu program dalam melakukan perhitungan, membandingkan nilai, mengolah kondisi, memberikan nilai, hingga memanipulasi data pada tingkat bit. Oleh karena itu, pemahaman mengenai jenis dan fungsi operator menjadi dasar yang penting dalam mempelajari pemrograman Python. Berikut operator pada python

 1. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi perhitungan matematika pada Python. Operator ini meliputi penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), modulus (%), perpangkatan (**), dan pembagian bulat (//).

 2. Operator Perbandingan

 Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan berupa nilai Boolean, yaitu True atau False. Contohnya adalah sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), dan lebih kecil atau sama dengan (<=).

 3. Operator Logika

 Operator logika digunakan untuk menggabungkan atau membalikkan suatu kondisi. Operator yang digunakan yaitu and, or, dan not. Operator ini biasanya digunakan dalam percabangan atau pemeriksaan kondisi.

 4. Operator Assignment

 Operator assignment digunakan untuk memberikan atau memperbarui nilai pada sebuah variabel. Operator yang umum digunakan adalah =, +=, -=, *=, /=, dan %=.

 5. Operator Bitwise

 Operator bitwise digunakan untuk melakukan operasi terhadap data dalam bentuk bit atau bilangan biner. Operator ini meliputi & (AND), | (OR), ^ (XOR), ~ (NOT), << (shift kiri), dan >> (shift kanan).

 6. Operator Lainnya

 Selain operator di atas, Python memiliki operator identitas dan keanggotaan. Operator identitas yaitu is dan is not digunakan untuk memeriksa apakah dua objek merupakan objek yang sama. Sementara itu, operator keanggotaan yaitu in dan not in digunakan untuk memeriksa apakah suatu nilai terdapat dalam sebuah data atau kumpulan.

Berdasarkan dasar teori yang telah dipelajari, dapat disimpulkan bahwa operator merupakan bagian penting dalam pemrograman Python yang digunakan untuk mengolah data dan variabel. Operator memiliki beberapa jenis dengan fungsi yang berbeda, seperti aritmatika untuk perhitungan, perbandingan dan logika untuk mengolah kondisi, assignment untuk mengatur nilai variabel, bitwise untuk operasi pada bit, serta operator lainnya untuk memeriksa identitas dan keanggotaan data. Dengan memahami penggunaan setiap operator, proses pembuatan dan pengolahan program Python dapat dilakukan dengan lebih mudah dan terstruktur.

### 2.2.1 Operator aritmatika "Claudya Annisa_2611533024"

```python
# Buat file dengan nama aritmatika_2611533024.py
# Buat program untuk operator aritmatika dalam Phyton
# Nama variabel ditambah 4 digit nim terakhir contoh: angka 1_3024
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_3024 = int(input("Masukkan angka-1: "))
angka2_3024 = int(input("Masukkan angka-2: "))

# Penjumlahan
hasil = angka1_3024 + angka2_3024
print ("\nOperator Penjumlahan")
print ("Hasil =", hasil)

# Pengurangan 
hasil = angka1_3024 - angka2_3024
print ("\nOperator Pengurangan")
print ("Hasil =",hasil)

# Perkalian
hasil = angka1_3024 * angka2_3024
print ("\nOperator Perkalian")
print ("Hasil =",hasil)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3024 !=0:
    hasil = angka1_3024 / angka2_3024
    print ("\nOperator Pembagian")    
    print ("Hasil =",hasil)

    hasil = angka1_3024 // angka2_3024
    print ("\nOperator Pembagian Bulat")
    print ("Hasil  =", hasil)

    hasil = angka1_3024 % angka2_3024
    print ("\nOperator Sisa Bagi")
    print ("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_3024 ** angka2_3024
print ("\nOperator Pangkat")
print ("Hasil =",hasil)
```

**Penjelasan :**
1. Operator Penjumlahan (+) :
Digunakan untuk menjumlahkan dua nilai. Pada program, operator ini digunakan untuk menjumlahkan angka1_3024 dan angka2_3024.
2. Operator Pengurangan (-) :
Digunakan untuk mengurangi suatu nilai dengan nilai lainnya. Program menggunakan operator ini untuk mengurangi angka1_3024 dengan angka2_3024.
3. Operator Perkalian (*) :
Digunakan untuk mengalikan dua nilai. Hasil perkalian angka1_3024 dan angka2_3024 kemudian ditampilkan.
4. Operator Pembagian (/) :
Digunakan untuk membagi satu nilai dengan nilai lainnya dan menghasilkan bilangan desimal jika diperlukan. Dalam program, pembagian dilakukan setelah memastikan angka kedua tidak bernilai 0.
5. Operator Pembagian Bulat (//) :
Digunakan untuk melakukan pembagian dan mengambil hasil bagi dalam bentuk bilangan bulat.
6. Operator Sisa Bagi (%) :
Digunakan untuk mendapatkan sisa dari hasil pembagian dua nilai. Operator ini sering digunakan untuk mengetahui apakah suatu bilangan habis dibagi atau tidak.
7. Operator Pangkat (**) :
Digunakan untuk menghitung perpangkatan. Pada program, angka1_3024 ** angka2_3024 berarti angka pertama dipangkatkan dengan angka kedua.

### 2.2.2 Operator Perbandingan "Claudya Annisa_2611533024"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_3024
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program ini menggunakan operator perbandingan dalam Phyton

angka1_3024 = int(input("input angka-1:"))
angka2_3024 = int(input("input angka-2:"))

# Lebih besar dari
hasil = angka1_3024 > angka2_3024
print("\nOperator Lebih Besar Dari")
print("angka1_3024 > angka2_3024 =", hasil)

# Lebih kecil dari
hasil = angka1_3024 < angka2_3024
print("\nOperator Lebih Kecil Dari")
print("angka1_3024 < angka2_3024 =", hasil)

# Lebih besar dari sama dengan
hasil = angka1_3024 >= angka2_3024
print("\nOperator lebih besar dari sama dengan")
print("angka1_3024 >= angka2_3024 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_3024 <= angka2_3024
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_3024 <= angka2_3024 =", hasil)

# Sama dengan
hasil = angka1_3024 == angka2_3024
print("\nOperator sama dengan")
print("angka1_3024 == angka2_3024 =", hasil)

# Tidak sama dengan
hasil = angka1_3024 != angka2_3024
print("\nOperator tidak sama dengan")
print("angka1_3024 != angka2_3024 =", hasil)

# Tambahan: perbandingan berantai dalam Python
hasil = 0 < angka1_3024 < 100
print("\nPerbandingan berantai")
print("0 < angka1_3024 < 100 =", hasil)

hasil = 0 < angka2_3024 < 100
print("0 < angka2_3024 < 100 =", hasil)
```

***PENJELASAN :***
1. Operator Lebih Besar (>) :
Digunakan untuk memeriksa apakah nilai pertama lebih besar daripada nilai kedua. Hasilnya berupa True jika kondisi benar dan False jika kondisi salah.
2. Operator Lebih Kecil (<) :
Digunakan untuk memeriksa apakah nilai pertama lebih kecil daripada nilai kedua. Hasil yang diberikan adalah True atau False.
3. Operator Lebih Besar atau Sama Dengan (>=) :
Digunakan untuk memeriksa apakah nilai pertama lebih besar atau sama dengan nilai kedua.
4. Operator Lebih Kecil atau Sama Dengan (<=) :
Digunakan untuk memeriksa apakah nilai pertama lebih kecil atau sama dengan nilai kedua.
5. Operator Sama Dengan (==) :
Digunakan untuk membandingkan apakah dua nilai memiliki nilai yang sama. Jika sama, hasilnya True, sedangkan jika berbeda hasilnya False.
6. Operator Tidak Sama Dengan (!=) :
Digunakan untuk memeriksa apakah dua nilai berbeda. Hasilnya True jika nilainya berbeda dan False jika nilainya sama.
7. Operator Perbandingan Berantai (0 < angka1_3024 < 100) :
Digunakan untuk memeriksa dua kondisi perbandingan sekaligus. Pada program ini, operator tersebut memeriksa apakah nilai angka1_3024 berada di antara 0 dan 100. Cara yang sama juga diterapkan pada angka2_3024.

### 2.2.3 Operator Assignment "Claudya Annisa_2611533024"

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 variabel nim terakhir contoh: angka1_3024
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Phyton

angka1_3024 = int(input("Input angka-1: "))
angka2_3024 = int(input("Input angka-2: "))

print("\nNilai awal angka1_3024 =",angka1_3024)
print("\nNilai awal angka2_3024 =", angka2_3024)

# Assignment biasa
hasil = angka1_3024 
print("\nAssigment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_3024
hasil += angka2_3024
print("\nAssigment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_3024
hasil -= angka2_3024
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_3024
hasil *= angka2_3024
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3024 !=0:
    hasil = angka1_3024
    hasil /= angka2_3024
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_3024
    hasil //= angka2_3024
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_3024
    hasil %= angka2_3024
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan,")
    print("Angka kedua tidak boleh bernilai 0.")

    # Operator tambahan: assignment perpangkatan
    hasil = angka1_3024
    hasil **= angka2_3024
    print("\nAssignment perpangkatan (**=)")
    print("Hasil =", hasil)
```

***PENJELASAN :***
1. Assignment biasa (=) :
Digunakan untuk memberikan nilai dari suatu variabel ke variabel lainnya. Pada program, hasil = angka1_3024 berarti nilai angka1_3024 disimpan ke dalam variabel hasil.
2. Assignment penambahan (+=)  :
Digunakan untuk menambahkan nilai ke variabel yang sudah ada. hasil += angka2_3024 memiliki arti hasil = hasil + angka2_3024.
3. Assignment pengurangan (-=) :
Digunakan untuk mengurangi nilai variabel dengan nilai tertentu. hasil -= angka2_3024 berarti hasil = hasil - angka2_3024.
4. Assignment perkalian (*=) :
Digunakan untuk mengalikan nilai yang tersimpan dalam variabel. hasil *= angka2_3024 berarti hasil = hasil * angka2_3024.
5. Assignment pembagian (/=) :
Digunakan untuk membagi nilai variabel dengan nilai lainnya. hasil /= angka2_3024 berarti hasil = hasil / angka2_3024.
6. Assignment pembagian bulat (//=) :
Digunakan untuk melakukan pembagian bulat dan menyimpan hasilnya langsung ke variabel. hasil //= angka2_3024 berarti hasil = hasil // angka2_3024.
7. Assignment sisa bagi (%=) :
Digunakan untuk mendapatkan sisa pembagian dan menyimpannya ke dalam variabel. hasil %= angka2_3024 berarti hasil = hasil % angka2_3024.
8. Assignment perpangkatan (**=) :
Digunakan untuk melakukan perpangkatan dan menyimpan hasilnya ke dalam variabel. hasil **= angka2_3024 berarti hasil = hasil ** angka2_3024.

### 2.2.4 Operator Lainnya "Claudya Annisa_2611533024"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_3024
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("\n===================================")
print("1. OPERATOR KEANGGOTAAN")
print("===================================")

# Input beberapa data yang dipisahkan dengan koma
input_data = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data = [int(angka.strip()) for angka in input_data.split(",")]

nilai_dicari = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil = nilai_dicari in data
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data, "=", hasil)

# Operator not in
hasil = nilai_dicari not in data
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data, "=", hasil)


print("\n===================================")
print("2. OPERATOR IDENTITAS")
print("===================================")

# objek1 menggunakan list dari input pengguna
objek1 = data

# objek2 merujuk pada objek yang sama dengan objek1
objek2 = objek1

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3 = data.copy()

print("objek1 =", objek1)
print("objek2 =", objek2)
print("objek3 =", objek3)

# Operator is
hasil = objek1 is objek2
print("\nOperator identitas IS")
print(objek1, "is", objek2, "=", hasil)

# Operator is not
hasil = objek1 is not objek3
print("\nOperator identitas IS NOT")
print(objek1, "is not", objek3, "=", hasil)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai:")
print("objek1 is objek3:", objek1 is objek3)
print("objek1 is objek3:", objek1 is objek3)
```

***PENJELASAN :***
1. Operator Keanggotaan in :
Digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah data, seperti list. Jika nilai ditemukan, hasilnya True, sedangkan jika tidak ditemukan hasilnya False.
2. Operator Keanggotaan not in :
Digunakan untuk memeriksa apakah suatu nilai tidak terdapat di dalam sebuah data. Hasilnya True jika nilai tidak ditemukan dan False jika nilai ditemukan.
3. Operator Identitas is :
Digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama. Pada program, objek2 = objek1, sehingga keduanya merujuk pada objek yang sama dan menghasilkan True.
4. Operator Identitas is not :
Digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang berbeda. objek3 dibuat menggunakan data.copy(), sehingga memiliki isi yang sama dengan objek1, tetapi merupakan objek yang berbeda. Oleh karena itu, objek1 is not objek3 menghasilkan True.
5. Perbedaan is dan == :
Operator is digunakan untuk membandingkan identitas objek, sedangkan == digunakan untuk membandingkan nilai atau isi objek. Jadi, dua objek dapat memiliki isi yang sama tetapi belum tentu merupakan objek yang sama.

### 2.2.5 Operator logika "Claudya Annisa_2611533024"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_3024
# Program ini menggunakan fungsi input()
# Program operator logika dalam Phyton

# Memasukkan nilai boolean 
# input tidak pekka terhadap huruf besar dan kecil
a1_3024 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3024 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_3024)
print("A2 =", a2_3024)

# Konjungsi: bernilai True jika keduanya True
hasil = a1_3024 and a2_3024
print("\nKonjungsi (AND)")
print("A1_3024 and A2_3024 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_3024 or a2_3024
print("\nDisjungsi (OR)")
print("A1_3024 or A2_3024 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_3024
print("\nNegasi A1 (NOT)")
print("not A1_3024 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_3024
print("\nNegasi A2 (NOT)")
print("not A2_3024 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_3024 != a2_3024
print("\nDisjungsi Eksklusif (XOR)")
print("A1_3024 XOR A2_3024 =", hasil)
```

***PENJELASAN :***
1. Operator AND (and) :
Digunakan untuk menggabungkan dua kondisi. Hasilnya True jika kedua nilai bernilai True. Jika salah satunya False, hasilnya False.
2. Operator OR (or) :
Digunakan untuk menggabungkan dua kondisi. Hasilnya True jika salah satu atau kedua nilai bernilai True. Hasilnya False jika keduanya False.
3. Operator NOT (not) :
Digunakan untuk membalikkan nilai Boolean. Jika nilai awal True, maka hasilnya menjadi False, dan sebaliknya. Pada program, operator ini digunakan untuk membalik nilai a1_3024 dan a2_3024.
4. Operator XOR (!=) :
Digunakan untuk menghasilkan True jika kedua nilai Boolean berbeda. Jika kedua nilai sama, hasilnya False. Pada program, XOR dibuat menggunakan operator !=.

### 2.2.6 Operator Bitwise "Claudya Annisa_2611533024"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_3024
# Program ini menggunakan fungsi input()

print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================")

angka1_3024 = int(input("Masukkan angka bitwise-1: "))
angka2_3024 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner:")
print("angka1_3024 =", angka1_3024, "| biner =", bin(angka1_3024))
print("angka2_3024 =", angka2_3024, "| biner =", bin(angka2_3024))

# Bitwise AND
hasil = angka1_3024 & angka2_3024
print("\nBitwise AND (&)")
print(angka1_3024, "&", angka2_3024, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise OR
hasil = angka1_3024 | angka2_3024
print("\nBitwise OR (|)")
print(angka1_3024, "|", angka2_3024, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise XOR
hasil = angka1_3024 ^ angka2_3024
print("\nBitwise XOR (^)")
print(angka1_3024, "^", angka2_3024, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise NOT
hasil = ~angka1_3024
print("\nBitwise NOT (~)")
print("~", angka1_3024, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kiri
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil = angka1_3024 << jumlah_geser
print("\nBitwise geser kiri (<<)")
print(angka1_3024, "<<", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kanan
hasil = angka1_3024 >> jumlah_geser
print("\nBitwise geser kanan (>>)")
print(angka1_3024, ">>", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))
```

***PENJELASAN :***
1. Bitwise AND (&) :
 Digunakan untuk membandingkan setiap bit dari dua bilangan. Hasilnya 1 jika kedua bit sama-sama 1, sedangkan jika salah satunya 0, hasilnya 0.
2. Bitwise OR (|) :
 Membandingkan setiap bit dari dua bilangan. Hasilnya 1 jika salah satu atau kedua bit bernilai 1. Jika keduanya 0, hasilnya 0.
3. Bitwise XOR (^) :
 Membandingkan setiap bit dari dua bilangan dan menghasilkan 1 jika kedua bit berbeda. Jika kedua bit sama, hasilnya 0.
4. Bitwise NOT (~) :
 Digunakan untuk membalik setiap bit dari suatu bilangan. Bit 1 menjadi 0 dan bit 0 menjadi 1. Dalam Python, hasil operator ~ juga mengikuti representasi bilangan bertanda sehingga hasil desimalnya dapat berupa bilangan negatif.
5. Bitwise Geser Kiri (<<) :
 Digunakan untuk menggeser bit ke arah kiri sebanyak jumlah posisi yang ditentukan. Pada bilangan bulat positif, setiap satu kali geser kiri secara umum setara dengan mengalikan nilai dengan 2.
6. Bitwise Geser Kanan (>>) :
 Digunakan untuk menggeser bit ke arah kanan sebanyak jumlah posisi yang ditentukan. Pada bilangan bulat positif, setiap satu kali geser kanan secara umum setara dengan membagi nilai dengan 2 dan mengambil hasil bulat.
7. Fungsi bin() dan format() :
bin() digunakan untuk menampilkan bilangan dalam bentuk biner, sedangkan format(hasil, "08b") digunakan untuk menampilkan hasil biner dengan panjang minimal 8 bit.
