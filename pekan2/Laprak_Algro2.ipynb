Tipe data merupakan klasifikasi atau jenis dari suatu nilai yang menentukan jenis operasi apa yang dapat dilakukan terhadap nilai tersebut, serta berapa besar ruang memori yang dibutuhkan untuk menyimpannya. Kedua konsep ini saling berkaitan erat, karena setiap variabel yang dideklarasikan dalam sebuah program harus memiliki tipe data tertentu agar komputer dapat mengetahui cara memperlakukan data yang disimpan di dalamnya.

## 2.1 Dasar Teori
Pada bab ini akan dijelaskan mengenai konsep-konsep dasar yang menjadi landasan dalam pelaksanaan praktikum "Tipe Data dan Variabel". Konsep-konsep tersebut meliputi pengertian variabel, konstanta, tipe data numerik (integer dan float), string, boolean, serta proses input dalam pemrograman. Pemahaman terhadap teori-teori dasar ini diperlukan sebagai acuan sebelum melakukan praktik langsung, sehingga mahasiswa dapat mengetahui konsep yang mendasari setiap proses yang dilakukan selama praktikum berlangsung.
 1 Variabel

Variabel adalah nama yang digunakan untuk merepresentasikan lokasi penyimpanan data di dalam memori komputer, yang nilainya dapat berubah-ubah selama proses eksekusi program berlangsung. Setiap variabel harus dideklarasikan terlebih dahulu dengan menentukan nama dan tipe datanya sebelum dapat digunakan dalam program.

 2. Konstanta

Konstanta adalah identitas penyimpan data yang serupa dengan variabel, namun memiliki nilai yang bersifat tetap dan tidak dapat diubah selama program dijalankan. Konstanta umumnya digunakan untuk menyimpan nilai baku yang tidak berubah, seperti nilai π (3.14159) atau batas maksimum suatu data.

 3. Numerik

Numerik merupakan kelompok tipe data yang digunakan untuk menyimpan nilai berupa angka, yang dapat diolah menggunakan operasi matematika seperti penjumlahan, pengurangan, perkalian, dan pembagian. Tipe data numerik terbagi menjadi dua kategori utama, yaitu integer dan float.

 4. Integer

Integer adalah tipe data numerik yang digunakan untuk menyimpan bilangan bulat, baik positif maupun negatif, tanpa mengandung nilai desimal.
Contoh: 10, -25, 1000.

 5. Float

Float (floating point) adalah tipe data numerik yang digunakan untuk menyimpan bilangan yang memiliki nilai desimal atau pecahan.
Contoh: 3.14, -0.5, 17.75.

 6.String

String adalah tipe data yang digunakan untuk menyimpan kumpulan karakter yang membentuk suatu kata, kalimat, atau teks. Penulisan data bertipe string umumnya diapit oleh tanda kutip.
Contoh: "Algoritma", "Universitas Andalas".

 7. Boolean

Boolean adalah tipe data yang hanya memiliki dua kemungkinan nilai, yaitu true (benar) atau false (salah). Tipe data ini banyak digunakan dalam proses pengambilan keputusan (percabangan) maupun perulangan pada suatu program.

 8.Input

Input adalah proses memasukkan data dari luar sistem, misalnya dari pengguna melalui keyboard, ke dalam suatu program. Data yang dimasukkan tersebut kemudian disimpan ke dalam variabel agar dapat diproses lebih lanjut oleh program.


Dengan memahami tipe data dan variabel, mahasiswa dapat menentukan jenis data yang sesuai serta menggunakannya dalam proses pengolahan data pada program Python.

nilai_3024 = 75 merupakan nilai mahasiswa, sedangkan batas_lulus_3024 = 75 merupakan nilai minimum yang harus dicapai agar dinyatakan lulus.
3. Menentukan status kelulusan
status_kelulusan = nilai_3024 >= batas_lulus_3024 digunakan untuk membandingkan nilai mahasiswa dengan batas kelulusan. Karena 75 ≥ 75, hasilnya adalah True, sehingga mahasiswa dinyatakan lulus.
Menampilkan hasil
4. Fungsi print() digunakan untuk menampilkan judul, nilai mahasiswa, dan status kelulusan ke layar.
5. Menggunakan operator logika and
Pada if is_lulus_3024 and is_cumlaude_3024, operator and memastikan kedua kondisi harus bernilai True. Karena keduanya True, program menampilkan ucapan bahwa mahasiswa lulus dengan predikat Cum Laude.
6. Kesimpulan
Program ini merupakan contoh penggunaan Boolean, operator perbandingan (>=), percabangan if, dan operator logika and dalam Python untuk menentukan status kelulusan mahasiswa.

### 2.2.2 Program Input "Claudya Annisa_2611533024"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_3024 = input('Masukkan nilai panjang: ') 
lebar_3024 =input('Masukkan nilai lebar: ')
luas_3024 = int(panjang_3024) * int(lebar_3024)
print("Luas=",luas_3024)
```

3. Menggunakan nama variabel sesuai NIM
Variabel panjang_3024, lebar_3024, dan luas_3024 menggunakan tambahan 4 digit terakhir NIM, yaitu 3024.
4. Mengubah tipe data menjadi integer
int(panjang_3024) dan int(lebar_3024) digunakan untuk mengubah data dari string menjadi integer (bilangan bulat) agar dapat dilakukan operasi perkalian.
5. Menghitung luas persegi panjang
luas_3024 = int(panjang_3024) * int(lebar_3024) menghitung luas dengan rumus:
Luas = panjang × lebar.

### 2.2.3 Program Konstanta "Claudya Annisa_2611533024"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3024 = float(input('Masukkan niali jari-jari: '))
luas_3024 = PI * jari_3024 * jari_3024
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f") % (jari_3024, luas_3024)
```

***PENJELASAN :***
1. Menggunakan konstanta PI
PI: Final = 3.14 digunakan untuk menyimpan nilai π (phi) sebesar 3,14. Final menandakan bahwa nilai tersebut dimaksudkan sebagai konstanta, yaitu nilai yang tidak seharusnya diubah selama program berjalan.
2. Menampilkan nilai PI
print("pi: %f" % (PI)) digunakan untuk menampilkan nilai PI. %f digunakan untuk menampilkan data berupa bilangan desimal.
3. Memasukkan nilai jari-jari
jari_3024 = float(input('Masukkan niali jari-jari: ')) digunakan untuk meminta pengguna memasukkan nilai jari-jari lingkaran. Fungsi float() digunakan agar nilai yang dimasukkan dapat berupa bilangan desimal.
4. Menghitung luas lingkaran
luas_3024 = PI * jari_3024 * jari_3024 digunakan untuk menghitung luas lingkaran dengan menggunakan nilai PI dan jari-jari.
5. Menampilkan hasil perhitungan
print() digunakan untuk menampilkan hasil luas lingkaran. Format %.2f digunakan agar angka ditampilkan dengan 2 angka di belakang koma.
6. Penggunaan variabel sesuai NIM
jari_3024 dan luas_3024 menggunakan tambahan 4 digit terakhir NIM, yaitu 3024, sesuai ketentuan tugas.
7. Kesimpulan
Program ini menunjukkan penggunaan konstanta, Final, input(), float(), operasi perkalian, dan print() dalam Python untuk menghitung luas lingkaran berdasarkan jari-jari yang dimasukkan pengguna.

### 2.2.4 Program Numerik"Claudya Annisa_2611533024"

```python
#Buat file dengan nama Numerik_2611533024.py
#nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variabel with numeric type value.
a_3024=100
print("The type of variabel having value",a_3024,"is",type(a_3024))

#create a variabel with float value.
b_3024=10.2345
print("The type of variabel having value",b_3024,"is",type(b_3024))

#create a variabel with complex value.
c_3024=100+3j
print("The type of variable having value",c_3024,"is",type(c_3024))
```

***PENJELASAN :***
1. Membuat variabel dengan tipe data integer
a_3024 = 100 digunakan untuk membuat variabel yang berisi bilangan bulat (integer). Nilai 100 tidak memiliki angka desimal.
2. Menampilkan tipe data integer
print(..., type(a_3024)) digunakan untuk menampilkan nilai variabel sekaligus mengetahui tipe datanya. Hasilnya adalah <class 'int'>, yang berarti variabel a_3024 bertipe integer.
3. Membuat variabel dengan tipe data float
b_3024 = 10.2345 digunakan untuk membuat variabel dengan nilai bilangan desimal (float).
4. Menampilkan tipe data float
print(..., type(b_3024)) digunakan untuk mengetahui tipe data dari b_3024. Hasilnya adalah <class 'float'>, yang berarti variabel tersebut bertipe float.
5. Membuat variabel dengan tipe data complex
c_3024 = 100 + 3j digunakan untuk membuat variabel dengan tipe data complex (bilangan kompleks). Dalam Python, huruf j menunjukkan bagian imajiner dari bilangan kompleks.
6. Menampilkan tipe data complex
print(..., type(c_3024)) digunakan untuk mengetahui tipe data dari c_3024. Hasilnya adalah <class 'complex'>, yang berarti variabel tersebut bertipe complex.
7. Penggunaan nama variabel sesuai NIM
Variabel a_3024, b_3024, dan c_3024 menggunakan tambahan 4 digit terakhir NIM, yaitu 3024, sesuai dengan ketentuan tugas.
8. Kesimpulan
Program ini menunjukkan tiga tipe data numerik yang ada dalam Python, yaitu integer (int) untuk bilangan bulat, float (float) untuk bilangan desimal, dan complex (complex) untuk bilangan kompleks. Fungsi type() digunakan untuk mengetahui tipe data dari setiap variabel.
