### 2.2.1 Program Jumlah Genap "Claudya Annisa_2611533024"

```python
# Buat file dengan nama jumlah_fenap_2611533024.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_3024
# Program ini menggunakan fungsi input()

ulang_3024 = int(input("Masukkan nilai batas: "))

jumlah = 0
for i_3024 in range(1, ulang_3024 +1):
    if i_3024 % 2 == 0:
        print(i_3024,end=" ")
        jumlah = jumlah + i_3024

        if i_3024 < ulang_3024:
            print("+", end=" ")
        else:
            print("=", jumlah, end="")
print()
print("Jumlah =:", jumlah)
```

Digunakan untuk membuat variabel jumlah dengan nilai awal 0 yang berfungsi menyimpan hasil penjumlahan bilangan genap.

Digunakan untuk melakukan perulangan dari angka 1 sampai nilai batas yang dimasukkan pengguna. Variabel i_3024 menyimpan nilai bilangan pada setiap perulangan.

4. if i_3024 % 2 == 0:

Digunakan untuk memeriksa apakah bilangan termasuk bilangan genap. Operator % menghitung sisa pembagian, sedangkan == 0 berarti sisa pembagiannya harus nol.

5. print(i_3024, end=" ")

Digunakan untuk menampilkan bilangan genap yang ditemukan. Parameter end=" " membuat bilangan ditampilkan dalam satu baris dan dipisahkan oleh spasi.

6. jumlah = jumlah + i_3024

Digunakan untuk menambahkan setiap bilangan genap ke variabel jumlah, sehingga diperoleh total penjumlahan seluruh bilangan genap.

7. if i_3024 < ulang_3024:

Digunakan untuk memeriksa apakah bilangan yang sedang diproses lebih kecil daripada nilai batas yang dimasukkan pengguna.

8. print("+", end=" ")

Digunakan untuk menampilkan tanda tambah (+) di antara bilangan genap agar hasilnya terlihat seperti operasi penjumlahan.

9. else:

Digunakan sebagai alternatif ketika kondisi pada if i_3024 < ulang_3024 tidak terpenuhi. Pada kode ini, bagian tersebut bertujuan menampilkan tanda sama dengan dan hasil penjumlahan.

10. print("=", jumlah, end="")

Digunakan untuk menampilkan tanda sama dengan (=) dan nilai total yang tersimpan dalam variabel jumlah.

Digunakan untuk membuat baris baru setelah proses perulangan selesai agar keluaran berikutnya ditampilkan pada baris selanjutnya.

Digunakan untuk menampilkan hasil akhir penjumlahan seluruh bilangan genap dengan keterangan Jumlah =:.

### 2.2.5 Program Nested For 4 "Claudya Annisa_2611533024"

```python
# Buat file dengan nama nested_for4_2611533024.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_3024
# Program ini menggunakan fungsi input()

tinggi_3024 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3024 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3024 = tinggi_3024
    c_3024 = a_3024
    lebar_3024= (2 * tinggi_3024) -2

    for i_3024 in range(1, tinggi_3024 + 1):
        b_3024 = c_3024 + 1

        for j_3024 in range(1, lebar_3024 +1):

            # Baris atas dan bawah
            if i_3024 == 1 or i_3024 == tinggi_3024:
                if j_3024 == 1 or j_3024 == lebar_3024:
                    print("#", end=" ")
                else:
                    print("=", end="") 

            # Baris isi
            else:
                if j_3024 == 1 or j_3024 == lebar_3024:
                    print("!", end=" ")
                else:
                    if j_3024 == c_3024:
                         print("<", end="")
                    elif j_3024 == b_3024:
                         print(">", end="")
                    elif j_3024 == (lebar_3024 - c_3024):
                         print("<", end="")
                    elif j_3024 == (lebar_3024 - c_3024 + 1):
                         print(">", end="")
                    elif j_3024 > b_3024 and j_3024 < (lebar_3024 -c_3024):
                         print("=", end="")
                    else:
                         print(" ",end="")

        print()

        # Logika asli Java
        a_3024 -= 2 
    
        if a_3024 <= 0:
            c_3024= (-a_3024) + 2
        else:
            c_3024 = a_3024
```

**PENJELASAN:**

Digunakan untuk menjalankan proses pembuatan pola apabila nilai tinggi yang dimasukkan merupakan bilangan genap.

Digunakan untuk menyimpan nilai tinggi ke variabel a_3024. Variabel ini nantinya digunakan untuk mengatur perubahan posisi pola pada setiap baris.

8. for i_3024 in range(1, tinggi_3024 + 1):
Digunakan sebagai perulangan luar (outer loop) untuk mengatur jumlah baris pola. Perulangan dimulai dari 1 hingga nilai tinggi yang dimasukkan pengguna.

Digunakan untuk menentukan posisi berikutnya setelah posisi yang disimpan dalam c_3024. Variabel ini membantu mengatur penempatan karakter < dan >.

10. for j_3024 in range(1, lebar_3024 + 1):
Digunakan sebagai perulangan dalam (inner loop) untuk memeriksa setiap kolom pada suatu baris. Perulangan ini menentukan karakter yang akan ditampilkan pada setiap posisi pola.

11. if i_3024 == 1 or i_3024 == tinggi_3024:
Digunakan untuk memeriksa apakah baris yang sedang diproses merupakan baris pertama atau baris terakhir. Kondisi or berarti cukup salah satu kondisi terpenuhi agar perintah di dalamnya dijalankan.

12. if j_3024 == 1 or j_3024 == lebar_3024:
Digunakan untuk memeriksa apakah posisi kolom berada di bagian paling kiri atau paling kanan pola pada baris pertama dan terakhir.

13. print("#", end=" ")
Digunakan untuk menampilkan karakter # pada kedua ujung baris pertama dan terakhir sebagai penanda sudut pola.

14. print("=", end="")
Digunakan untuk menampilkan karakter = pada bagian tengah baris pertama dan terakhir. Karakter tersebut membentuk garis horizontal pada pola.

15. else: pada bagian baris isi
Digunakan ketika baris yang sedang diproses bukan baris pertama maupun baris terakhir. Pada bagian ini, program mulai membentuk sisi dan pola bagian dalam.

16. if j_3024 == 1 or j_3024 == lebar_3024:
Digunakan untuk memeriksa apakah kolom berada di sisi paling kiri atau paling kanan pada baris isi.

17. print("!", end=" ")
Digunakan untuk menampilkan karakter ! pada sisi kiri dan kanan pola sehingga membentuk batas vertikal.

18. if j_3024 == c_3024:
Digunakan untuk memeriksa apakah posisi kolom sama dengan nilai c_3024. Jika sesuai, program menampilkan karakter < pada posisi tersebut.

19. elif j_3024 == b_3024:
Digunakan untuk memeriksa posisi kolom berikutnya, yaitu nilai b_3024. Jika kondisi terpenuhi, program menampilkan karakter >.

20. elif j_3024 == (lebar_3024 - c_3024):
Digunakan untuk menentukan posisi karakter < berdasarkan lebar pola dikurangi nilai c_3024. Kondisi ini membantu membentuk pola pada bagian dalam sisi lainnya.

21. elif j_3024 == (lebar_3024 - c_3024 + 1):
Digunakan untuk menentukan posisi karakter > yang berdekatan dengan posisi karakter sebelumnya pada sisi lainnya.

22. elif j_3024 > b_3024 and j_3024 < (lebar_3024 - c_3024):
Digunakan untuk memeriksa apakah posisi kolom berada di antara batas posisi yang telah ditentukan. Jika benar, program menampilkan karakter = untuk mengisi bagian dalam pola.

23. else: dan print(" ", end="")

### 2.2.8 Program Perulangan For 3 "Claudya Annisa_2611533024"

```python
# Buat file dengan nama perulangan_for3_2611533024.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_3024
# Program ini menggunakan fungsi input()

ulang_3024 = int(input("Masukkan jumlah perulangan: "))

jumlah = 0
for i in range(1, ulang_3024 + 1):
    print(i,end=" ")
    jumlah = jumlah + i

    if i < ulang_3024:
        print(" + ", end="")
    else:
        print(" = ", jumlah,end="")
print()
print("Jumlah =", jumlah)
```

**PENJELASAN:**

1. ulang_3024 = int(input("Masukkan jumlah perulangan: "))
Digunakan untuk meminta pengguna memasukkan jumlah perulangan yang diinginkan. Fungsi input() menerima masukan, sedangkan int() mengubah masukan menjadi bilangan bulat yang disimpan dalam variabel ulang_3024.

2. jumlah = 0
Digunakan untuk membuat variabel jumlah dengan nilai awal 0. Variabel ini berfungsi untuk menyimpan hasil penjumlahan seluruh angka selama perulangan berlangsung.

3. for i in range(1, ulang_3024 + 1):
Digunakan untuk menjalankan perulangan dari angka 1 hingga nilai yang dimasukkan pengguna. Penambahan + 1 memastikan angka batas terakhir ikut diproses.

4. print(i, end=" ")
Digunakan untuk menampilkan angka yang sedang diproses dalam perulangan. Parameter end=" " membuat angka ditampilkan dalam satu baris dan dipisahkan oleh spasi.

5. jumlah = jumlah + i
Digunakan untuk menambahkan nilai i ke variabel jumlah pada setiap perulangan. Dengan demikian, total penjumlahan akan terus bertambah hingga perulangan selesai.

6. if i < ulang_3024:
Digunakan untuk memeriksa apakah angka yang sedang diproses masih lebih kecil daripada batas perulangan. Jika kondisi benar, program akan menampilkan tanda tambah.

7. print(" + ", end="")
Digunakan untuk menampilkan tanda tambah (+) di antara angka-angka yang dijumlahkan. Parameter end="" membuat keluaran tetap berada pada baris yang sama.

8. else:
Digunakan ketika kondisi i < ulang_3024 tidak terpenuhi, yaitu saat perulangan mencapai angka terakhir. Bagian ini digunakan untuk menampilkan tanda sama dengan dan hasil penjumlahan.

9. print(" = ", jumlah, end="")
Digunakan untuk menampilkan tanda sama dengan (=) diikuti hasil penjumlahan yang tersimpan dalam variabel jumlah.

10. print()
Digunakan untuk membuat baris baru setelah proses perulangan selesai sehingga keluaran berikutnya tampil pada baris berbeda.

11. print("Jumlah =", jumlah)
Digunakan untuk menampilkan hasil akhir penjumlahan seluruh angka dengan keterangan Jumlah =.

### 2.2.8 Program Latihan 5 "Claudya Annisa_2611533024"

```python
tinggi_3024 = int(input("Masukkan tinggi segitiga: "))

for i_3024 in range(1, tinggi_3024 + 1):
    for j_3024 in range(tinggi_3024 - i_3024):
        print(" ", end="")

    for k_3024 in range(i_3024):
        print("* ", end="")

    print()
```

**PENJELASAN:**

Digunakan sebagai perulangan luar yang mengatur jumlah baris segitiga. Perulangan dimulai dari angka 1 hingga nilai tinggi yang dimasukkan pengguna.

3. for j_3024 in range(tinggi_3024 - i_3024):
Digunakan untuk mencetak spasi sebelum bintang pada setiap baris. Jumlah spasi semakin berkurang ketika baris bertambah, sehingga bentuk segitiga terlihat berada di tengah.

4. print(" ", end="")
Digunakan untuk menampilkan satu spasi tanpa berpindah ke baris baru. Perintah ini membantu mengatur posisi bintang agar membentuk segitiga.

5. for k_3024 in range(i_3024):
Digunakan untuk mengatur jumlah bintang yang dicetak pada setiap baris. Jumlah bintang mengikuti nilai i_3024, sehingga semakin ke bawah, semakin banyak bintang yang ditampilkan.

6. print("* ", end="")
Digunakan untuk mencetak bintang dengan satu spasi setelahnya. Parameter end="" membuat bintang tetap berada pada baris yang sama.

7. print()
Digunakan untuk berpindah ke baris berikutnya setelah spasi dan bintang selesai dicetak. Dengan demikian, setiap baris membentuk susunan segitiga yang teratur.
