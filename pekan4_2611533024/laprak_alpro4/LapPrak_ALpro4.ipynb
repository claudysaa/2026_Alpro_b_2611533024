## 2.1 Dasar Teori
Conditional merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Dengan conditional, program dapat menentukan perintah yang akan dijalankan sesuai dengan nilai atau keadaan yang diberikan.

Dalam Python, conditional dapat digunakan dalam beberapa bentuk, yaitu if, if-elif-else, multi if, match case, dan nilai dalam conditional. Penggunaan conditional membuat program lebih dinamis karena dapat menghasilkan keluaran yang berbeda sesuai dengan kondisi yang diberikan.

2.2.1 Nilai dalam Conditional

Nilai merupakan data yang digunakan sebagai dasar untuk menentukan suatu kondisi dalam program. Dalam conditional, nilai biasanya dibandingkan dengan nilai tertentu menggunakan operator perbandingan seperti >, <, >=, <=, ==, dan !=.

2.2.2 If

if digunakan untuk menjalankan perintah apabila suatu kondisi bernilai benar atau True.

2.2.3 If-Elif-Else

if-elif-else digunakan untuk memeriksa beberapa kondisi. Program akan memeriksa kondisi mulai dari if, kemudian elif, dan menjalankan else jika tidak ada kondisi sebelumnya yang terpenuhi.

2.2.4 Multi If

Multi if merupakan penggunaan beberapa pernyataan if secara terpisah. Setiap kondisi diperiksa secara independen sehingga lebih dari satu kondisi dapat dijalankan apabila semuanya terpenuhi.

2.2.5 Match Case

match case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan yang telah ditentukan. Setiap pilihan ditulis menggunakan case.

### 2.2.4 Program multi_if2 "Claudya Annisa_2611533024"

```python
# Buat file dengan nama multi_else1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_3024
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3024 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah umur mengetik 'y' atau 'ya')
input_member_3024 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_3024 = input_member_3024 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3024 = ("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3024 = input_promo_3024 in ["y", "ya"]

total_diskon_persen_3024 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3024 > 1000000:
    total_diskon_persen_3024 += 10 # Diskon belanja besar

if is_member_3024:
    total_diskon_persen_3024 += 5 # Diskon member

if kode_promo_valid_3024:
    total_diskon_persen_3024 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3024 = total_belanja_3024 * (total_diskon_persen_3024 / 100)
total_bayar_3024 = total_belanja_3024 - nominal_diskon_3024

#Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon_3024  : {total_diskon_persen_3024}% (Rp{nominal_diskon_3024:,.0f})")
print(f"Total Bayar_3024   : Rp {total_bayar_3024:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_3024}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**PENJELASAN :**

1. **Input Total Belanja**
   Program meminta pengguna memasukkan total belanja. Nilai tersebut dapat berupa angka desimal.

2. **Input Status Member**
   Program meminta pengguna menentukan apakah merupakan member atau bukan dengan pilihan `y` atau `t`. Input kemudian diperiksa untuk menentukan status member.

3. **Input Kode Promo**
   Program meminta pengguna memasukkan status kode promo. Jika kode promo dinyatakan valid, pengguna akan mendapatkan tambahan diskon.

4. **Menentukan Diskon Awal**
   Total diskon awal ditetapkan sebesar **0%** sebelum program melakukan pengecekan terhadap setiap kondisi.

5. **Diskon Belanja Besar**
   Jika total belanja lebih dari **Rp1.000.000**, pengguna mendapatkan tambahan diskon sebesar **10%**.

6. **Diskon Member**
   Jika pengguna merupakan member, maka mendapatkan tambahan diskon sebesar **5%**.

7. **Diskon Kode Promo**
   Jika kode promo valid, pengguna mendapatkan tambahan diskon sebesar **15%**.

8. **Akumulasi Diskon**
   Setiap kondisi diperiksa secara terpisah. Jika beberapa kondisi terpenuhi sekaligus, diskon akan **ditambahkan atau diakumulasikan**. Misalnya, jika ketiga kondisi terpenuhi, total diskon menjadi **30%**.

9. **Menghitung Nominal Diskon**
   Program menghitung jumlah uang yang mendapatkan potongan berdasarkan total belanja dan persentase diskon.

10. **Menghitung Total Bayar**
    Setelah nominal diskon diketahui, program mengurangi total belanja dengan nominal diskon untuk mendapatkan jumlah yang harus dibayar.

11. **Menampilkan Hasil**
    Program menampilkan rincian pembayaran berupa **total persentase diskon, nominal diskon, dan total yang harus dibayar**.

12. **Konsep Multi-If**
    Program menggunakan beberapa kondisi `if` secara terpisah sehingga setiap kondisi dapat diperiksa secara independen dan beberapa diskon dapat berlaku secara bersamaan.


### 2.2.5 Program match_case1 "Claudya Annisa_2611533024"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3024
# Program ini menggunakan hasil input()
# Program konversi angka menjadi nama bulan

bulan = int(input("Masukkan angka bulan (1 - 12): "))

match bulan:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**PENJELASAN :**

1. **Input Angka Bulan**
   Program meminta pengguna memasukkan angka bulan dari **1 sampai 12**. Input kemudian diubah menjadi bilangan bulat menggunakan `int()`.

2. **Penggunaan `match case`**
   Program menggunakan `match case` untuk mencocokkan angka yang dimasukkan dengan pilihan bulan yang tersedia.

3. **Case 1–12**
   Setiap `case` mewakili satu bulan. Contohnya, angka 1 menghasilkan **Januari**, angka 2 menghasilkan **Februari**, dan seterusnya sampai angka 12 yang menghasilkan **Desember**.

4. **Case `_`**
   `case _` digunakan sebagai kondisi selain pilihan yang tersedia. Jika pengguna memasukkan angka di luar **1–12**, program akan menampilkan **“Angka tidak valid”**.

5. **Tujuan Program**
   Program ini bertujuan untuk mengonversi angka yang dimasukkan pengguna menjadi **nama bulan** yang sesuai.

6. **Kesimpulan**
   Program menunjukkan penggunaan `match case` untuk memilih satu hasil berdasarkan nilai input yang diberikan. Ini membuat pilihan berdasarkan angka menjadi lebih terstruktur dan mudah dibaca.


### 2.2.6 Program if_elif_else1 "Claudya Annisa_2611533024"

```python
# Buat nama dengan nama if_elif_else1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3024
# Program ini menggunakan fungsi input()

umur_3024 = int(input("Input umur anda: "))
sim_3024 = input("Apakah Anda Sudah Punya Sim C ")[0]

if umur_3024 >= 17 and sim_3024 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_3024 >= 17 and sim_3024 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_3024 < 17 and sim_3024 == 'y':
    print("Anda belum cukup umur punya sim")
else:
    print("Anda belum  cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**PENJELASAN :**

1. **Input Umur**
   Program meminta pengguna memasukkan umur. Nilai umur diubah menjadi bilangan bulat menggunakan `int()`.

2. **Input Status SIM C**
   Program meminta pengguna memasukkan apakah sudah memiliki SIM C. Karakter pertama dari input digunakan sebagai jawaban, yaitu `y` untuk ya.

3. **Kondisi Pertama (`if`)**
   Jika umur **17 tahun atau lebih** dan sudah memiliki SIM C, program menampilkan bahwa pengguna sudah dewasa dan boleh membawa motor.

4. **Kondisi Kedua (`elif`)**
   Jika umur **17 tahun atau lebih**, tetapi belum memiliki SIM C, program menampilkan bahwa pengguna sudah dewasa tetapi tidak boleh membawa motor.

5. **Kondisi Ketiga (`elif`)**
   Jika umur **di bawah 17 tahun**, tetapi pengguna memasukkan bahwa sudah memiliki SIM C, program menampilkan bahwa pengguna belum cukup umur untuk memiliki SIM.

6. **Kondisi Terakhir (`else`)**
   Jika semua kondisi sebelumnya tidak terpenuhi, program menampilkan bahwa pengguna belum cukup umur dan tidak boleh membawa motor.

7. **Program Selesai**
   Setelah salah satu kondisi dijalankan, program menampilkan **“Program Selesai”**.

8. **Konsep `if-elif-else`**
   Program menggunakan `if-elif-else` untuk memeriksa beberapa kondisi secara berurutan. Setelah menemukan kondisi yang terpenuhi, program menjalankan perintah pada kondisi tersebut dan tidak memeriksa kondisi berikutnya.

```python
# Program Sistem Loket Alpro Adventure Park
# NIM: 2611533024

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Input Data Pengunjung
nama_pengunjung_3024 = input("Masukkan Nama Pengunjung: ")
umur_3024 = int(input("Input umur anda: "))
sim_3024 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()
jumlah_tiket_3024 = int(input("Masukkan jumlah tiket        : "))

# Validasi jumlah tiket dengan IF tunggal
if jumlah_tiket_3024 <= 0:
    print("Peringatan: Jumlah tiket tidak valid.")

# 2. Pilihan Paket Wahana
print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba       (Rp 50.000)")
print("2. Arung Jeram        (Rp 75.000)")
print("3. Motor ATV Ekstrim  (Rp 120.000)")
print("4. Roller Coaster Kilat (Rp 100.000)")
print("5. All-Access VIP     (Rp 220.000)")

paket_3024 = int(input("Masukkan nomor paket (1-5): "))

harga_satuan_3024 = 0
nama_paket_3024 = ""
paket_valid_3024 = True

# Match-case
match paket_3024:
    case 1:
        nama_paket_3024 = "Safari Rimba"
        harga_satuan_3024 = 50000
    case 2:
        nama_paket_3024 = "Arung Jeram"
        harga_satuan_3024 = 75000
    case 3:
        nama_paket_3024 = "Motor ATV Ekstrim"
        harga_satuan_3024 = 120000
    case 4:
        nama_paket_3024 = "Roller Coaster Kilat"
        harga_satuan_3024 = 100000
    case 5:
        nama_paket_3024 = "All-Access VIP"
        harga_satuan_3024 = 220000
    case _:
        print("Paket wahana tidak valid!")
        paket_valid_3024 = False

if not paket_valid_3024:
    raise SystemExit

# Input member dan promo
is_member_3024 = input("Apakah Anda member? (y/t): ").strip().lower()
kode_promo_valid_3024 = input("Apakah kode promo valid? (y/t): ").strip().lower()

# 3. Validasi izin berkendara
print("\n--- Validasi Izin Wahana ---")

if paket_3024 == 3 and umur_3024 >= 17 and sim_3024 == 'y':
    print("Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_3024 == 3 and umur_3024 >= 17 and sim_3024 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_3024 == 3 and umur_3024 < 17 and sim_3024 == 'y':
    print("Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_3024 == 3 and umur_3024 < 17:
    print("Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif paket_3024 != 3 and umur_3024 >= 10:
    print("Anda memenuhi batas umur untuk wahana.")
else:
    print("Anda belum cukup umur untuk wahana ini.")

# 4. Menghitung subtotal
subtotal_3024 = harga_satuan_3024 * jumlah_tiket_3024
total_diskon_persen_3024 = 0

# Multi-if untuk akumulasi diskon
if subtotal_3024 >= 200000:
    total_diskon_persen_3024 += 10

if is_member_3024 in ['y', 'ya']:
    total_diskon_persen_3024 += 5

if kode_promo_valid_3024 in ['y', 'ya']:
    total_diskon_persen_3024 += 15

if jumlah_tiket_3024 >= 5:
    total_diskon_persen_3024 += 5

# 5. Menghitung pembayaran
nominal_diskon_3024 = subtotal_3024 * (total_diskon_persen_3024 / 100)
total_bayar_3024 = subtotal_3024 - nominal_diskon_3024

# Evaluasi bonus dengan IF-ELSE
if total_bayar_3024 > 300000:
    bonus_3024 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    bonus_3024 = "Terima kasih telah berkunjung."

# 6. Rincian pembayaran
print("\n=== RINCIAN PEMBAYARAN ===")
print(f"Nama Pengunjung     : {nama_pengunjung_3024}")
print(f"Paket Wahana        : {nama_paket_3024}")
print(f"Harga Satuan        : Rp {harga_satuan_3024:,.0f}")
print(f"Jumlah Tiket        : {jumlah_tiket_3024}")
print(f"Subtotal            : Rp {subtotal_3024:,.0f}")
print(f"Total Diskon        : {total_diskon_persen_3024}%")
print(f"Nominal Diskon      : Rp {nominal_diskon_3024:,.0f}")
print(f"Total Bayar         : Rp {total_bayar_3024:,.0f}")
print(f"Bonus               : {bonus_3024}")
print("=== TRANSAKSI SELESAI ===")
```
